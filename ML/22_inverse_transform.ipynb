# Inverse Transform / Encoding 
The inverse transform or encoding is the process of converting the transformed data back to its original form. This is typically done after applying a transformation to the data, 
- such as scaling,
- normalization,
- or encoding categorical variables. 
  The inverse transform allows us to interpret the results in the context of the original data.

- Label Encoding
- ordinal encoding
- one-hot encoding
- binary encoding

```python
#import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split 
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.preprocessing import LabelEncoder, StandardScaler, OneHotEncoder , OrdinalEncoder
```

```python
# load the data
df = sns.load_dataset('titanic')
df.head()
```

```python
# impute missing values age, embarked, embarke_town, deck
df['age'] = df['age'].fillna(df['age'].mean())
df['embarked'] = df['embarked'].fillna(df['embarked'].mode
()[0])
df['embark_town'] = df['embark_town'].fillna(df['embark_town'].mode()[0])
df.drop('deck', axis=1, inplace=True)
```

```python
df.isnull().sum()
```

```python
# ecode the categorical variables
le_sex = LabelEncoder()
le_embarked = LabelEncoder()
le_embark_town = LabelEncoder()
le_who = LabelEncoder()
le_adult_male = LabelEncoder()
le_alive = LabelEncoder()
le_alone = LabelEncoder()
le_class = LabelEncoder()


df['sex'] = le_sex.fit_transform(df['sex'])
df['embarked'] = le_embarked.fit_transform(df['embarked'])
df['embark_town'] = le_embark_town.fit_transform(df['embark_town'])
df['who'] = le_who.fit_transform(df['who'])
df['adult_male'] = le_adult_male.fit_transform(df['adult_male'])
df['alive'] = le_alive.fit_transform(df['alive'])
df['alone'] = le_alone.fit_transform(df['alone'])
df['class'] = le_class.fit_transform(df['class'])
df.head()
```

```python
# inverse transform the encoded variables
df['sex']=le_sex.inverse_transform(df['sex'])
df['embarked']=le_embarked.inverse_transform(df['embarked'])
df['embark_town']=le_embark_town.inverse_transform(df['embark_town'])   
df['who']=le_who.inverse_transform(df['who'])
df['adult_male']=le_adult_male.inverse_transform(df['adult_male'])
df['alive']=le_alive.inverse_transform(df['alive'])
df['alone']=le_alone.inverse_transform(df['alone'])
df['class']=le_class.inverse_transform(df['class'])

df.head()
```
