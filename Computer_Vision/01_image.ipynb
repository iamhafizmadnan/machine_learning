```python
#import libraries
import numpy as np
import cv2 as cv
```

```python
# read the image
img = cv.imread("E:/Machine Learning/Computer_Vision/images/adnan.png")
```

```python
# resize the image
img = cv.resize(img, (400, 400))
cv.imshow("Image", img)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# convert the image to grayscale
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
cv.imshow("Grayscale Image", gray)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# read webcam using opencv
cap = cv.VideoCapture(0)
while True:
    ret, frame = cap.read()
    if not ret:
        break
    cv.imshow("Webcam", frame)
    if cv.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv.destroyAllWindows()
```
